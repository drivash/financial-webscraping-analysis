**LIMPIEZA DE LOS DATOS DE LAS ACCIONES**

Para evitar errores en la ejecución del programa de web scraping para la obtención de datos de las acciones, se hicieron dos ejecuciones independientes, obteniendo los datos de 50 acciones en cada una de ellas. Por ello, se tienen 2 Excel iniciales. 
Antes de prodeder con la limpieza de datos, se ha modificado una celda (un valor de una acción) con un dato erróneo cuya causa es desconocida.

La limpieza de los datos de las acciones es sencilla y requiere de pocos pasos. Estos son:

* Leer los dos dataframe

* Concatenar ambos usando 'ignore_index = True' para respetar el índice consecutivo de cada acción.

* Eliminar la primera fila, dado que contiene valores incorrectos.

* En todas las columnas que contienen valores de acciones, se cambian los puntos por comas, para así poder reformatear las celdas a tipo 'float'.

Por último, convertimos el dataframe a un excel con nombre 'Datos_Acciones'.

```python
import pandas as pd
```

```python
# Leemos los dos archvos procedentes del Web Scrapping
df1 = pd.read_excel('../data/Datos_Acciones1.xlsx', header=0, index_col=None)
df2 = pd.read_excel('../data/Datos_Acciones2.xlsx', header=0, index_col=None)

# Los concatenamos para tener uno solo
df = pd.concat([df1, df2], ignore_index=True)

df = df.iloc[1:]

# Cambiamos los valores de las celdas. primero sustituimos las , por . y luego establecemos el tipo de datos como numérico
df.iloc[:, 4:] = df.iloc[:, 4:].apply(lambda x: x.str.replace(',', '.').astype(float))

df.to_excel('../data/Datos_Acciones.xlsx', sheet_name='Datos_Acciones', index= False)
```
